# 04 · EDA longitudinal y asociaciones por ronda

**Objetivo:** comparar evoluciones sin sumar personas en distintas rondas y descubrir asociaciones candidatas a análisis predictivo. **Pendiente CP2**: matrices Spearman y Cramér V para cada año, comparativas BrBG y evaluación de attrición.


```python
from tesis_enasem.config import RAW_DIR, ROUND_TO_YEAR
from tesis_enasem.io import load_stata
from tesis_enasem.panel import add_year, filter_population
from tesis_enasem.missing import numeric_enasem
import pandas as pd

df=filter_population(add_year(load_stata(RAW_DIR/'simpleMHAS.dta',preserve_missing=True),ROUND_TO_YEAR))
continuas=['edad','educacion','n_enf','n_sint_depr','n_abvd','n_aivd','n_mov','recuerdo1','recuerdo2','imp_neto']
for col in continuas: df[col]=numeric_enasem(df[col])
```

## Correlación de Spearman por ronda (no acumulada)


```python
CORRELACION_ANO=2018
subset=df[df.ano.eq(CORRELACION_ANO)]
corr=subset[continuas].corr(method='spearman')
display(corr.round(2))
```
